```python
#字符串
import re

def is_palindrome(s):
    # 使用正则表达式只保留字母和数字，并转换为小写
    cleaned = re.sub(r'[^a-zA-Z0-9]', '', s).lower()
    
    # 判断清洗后的字符串是否等于其反转
    return cleaned == cleaned[::-1]

def is_abecedarian(s):
    # 移除空格并转小写以便比较
    cleaned = s.replace(" ", "").lower()
    # 遍历字符串，检查当前字符是否小于前一个字符
    for i in range(1, len(cleaned)):
        if cleaned[i] < cleaned[i-1]:
            return False   
    return True

def remove_vowels(s):
    # 定义元音字母集合（包含大小写）
    vowels = "aeiouAEIOU"
    result = ""
    
    # 遍历并保留非元音字符
    for char in s:
        if char not in vowels:
            result += char
            
    return result

# --- 测试代码 ---
if __name__ == "__main__":
    # 测试 1: is_palindrome
    print("--- Test 1: Palindrome ---")
    print(is_palindrome("level"))
    print(is_palindrome("rats live on no evil star"))
    print(is_palindrome("T A C O cat"))
    print(is_palindrome("hello"))

    # 测试 2: is_abecedarian
    print("\n--- Test 2: Abecedarian ---")
    print(is_abecedarian("adder"))
    print(is_abecedarian("beet"))
    print(is_abecedarian("dog"))
    print(is_abecedarian("abcd efgh xyz"))

    # 测试 3: remove_vowels
    print("\n--- Test 3: Remove Vowels ---")
    print(remove_vowels('cat'))
    print(remove_vowels('audio'))
    print(remove_vowels('bcdfg'))
```

```python
#列表
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是一个列表")
    return lst[::-1]

def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False# 如果发现前一个元素大于后一个元素，说明不是升序，直接返回 False
    return True # 如果遍历完都没有发现逆序的情况，说明是升序，返回 True

# 测试 list_reverse
print(list_reverse([1, 2, 3])) 

# 测试 is_sorted
print(is_sorted([1, 2, 3]))     
print(is_sorted([1, 3, 2]))     
```

```python
#元组
def vec_scalar_mult(t, s):
    """向量标量乘法：返回 t 中每个元素乘以 s 的新元组"""
    # 检查输入类型：t 必须是元组，s 必须是数字(排除布尔)，t 的元素也必须是数字
    if (not isinstance(t, tuple) or 
        not isinstance(s, (int, float)) or isinstance(s, bool) or
        not all(isinstance(x, (int, float)) and not isinstance(x, bool) for x in t)):
        raise TypeError("输入类型错误：t应为数字元组，s应为数字")
    return tuple(x * s for x in t)

def vec_inner_product(t1, t2):
    """向量内积：返回两个向量内积的浮点数"""
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("输入必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个元组长度必须一致")      
    # 使用 zip 配对元素，sum 求和，最后转为 float
    return float(sum(x * y for x, y in zip(t1, t2)))

def check_valid_mx(mx):
    """检查矩阵有效性：外层是元组，内层是等长元组，元素是数字"""
    if not isinstance(mx, tuple):
        return False
 
    row_len = None
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if not all(isinstance(x, (int, float)) and not isinstance(x, bool) for x in row):
            return False
         
        # 记录第一行长度，后续行必须与此长度相同
        if row_len is None:
            row_len = len(row)
        elif len(row) != row_len:
            return False
    return True

def mx_vec_mult(mx, vec):
    """矩阵乘向量：返回结果元组"""
    # 检查矩阵和向量的合法性
    if not check_valid_mx(mx):
        raise TypeError("第一个参数必须是有效的数值矩阵（元组的元组）")
    if not isinstance(vec, tuple):
        raise TypeError("第二个参数必须是向量（元组）")
    if len(mx[0]) != len(vec):
        raise ValueError("矩阵列数必须等于向量长度") 
    # 矩阵乘法逻辑：每一行与向量做内积
    result = []
    for row in mx:
        val = sum(r * v for r, v in zip(row, vec))
        result.append(val)
    return tuple(result)

if __name__ == "__main__":
    print(vec_scalar_mult((1, 2, 3), 2))     
    print(vec_inner_product((1, 2, 3), (4, 5, 6))) 
    valid_matrix = ((1, 2), (3, 4))
    invalid_matrix = ((1, 2), (3,))
    print(check_valid_mx(valid_matrix))       
    print(check_valid_mx(invalid_matrix))     
    print(mx_vec_mult(((1, 2), (3, 4)), (5, 6))) 
```

```python
#更多向量
import functools
@functools.total_ordering
class Vector:
    def __init__(self, dimension, elements=None):
        # 1. 维度校验
        if not isinstance(dimension, int) or isinstance(dimension, bool) or dimension < 0:
            raise ValueError("维度必须是非负整数")
        self.dimension = dimension
        # 2. 元素校验与默认处理
        if elements is None:
            self.elements = (0,) * dimension
        else:
            if not isinstance(elements, (list, tuple)) or len(elements) != dimension:
                raise ValueError(f"元素个数必须与维度({dimension})一致")
            if not all(isinstance(e, (int, float)) and not isinstance(e, bool) for e in elements):
                raise TypeError("元素必须是整数或浮点数")
            self.elements = tuple(elements)
    # 3. 比较运算符
    def __eq__(self, other):
        if not isinstance(other, Vector): return NotImplemented
        return self.dimension == other.dimension and self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector): return NotImplemented
        if self.dimension != other.dimension:
            raise ValueError("不能比较维度不同的向量")
        return self.elements < other.elements  # 利用元组自带的字典序比较

    # 4. 内积方法
    def dot(self, other):
        if not isinstance(other, Vector) or self.dimension != other.dimension:
            raise ValueError("参数必须是维度相同的 Vector 对象")
        return sum(x * y for x, y in zip(self.elements, other.elements))

    # 5. 乘法运算符
    def __mul__(self, other):
        if isinstance(other, (int, float)) and not isinstance(other, bool):
            return Vector(self.dimension, tuple(e * other for e in self.elements))
        if isinstance(other, Vector):
            if self.dimension != other.dimension:
                raise ValueError("向量相乘要求维度一致")
            return Vector(self.dimension, tuple(x * y for x, y in zip(self.elements, other.elements)))
        return NotImplemented

    def __rmul__(self, other):
        return self.__mul__(other)  # 处理 2 * v 的情况

    def __repr__(self):
        return f"Vector({self.dimension}, {self.elements})"

if __name__ == "__main__":
    # 基础测试
    v1 = Vector(2, (2, 4))
    v2 = Vector(2, (2, 5))
    v_zero = Vector(3)
    print(f"v1: {v1}, v2: {v2}, 默认零向量: {v_zero}")

    # 比较测试
    print(f"v1 == v2: {v1 == v2}")  
    print(f"v1 < v2: {v1 < v2}")    
    print(f"v1 <= v2: {v1 <= v2}")  

    # 内积测试
    print(f"v1 · v2 = {v1.dot(v2)}") # 24

    # 乘法测试
    print(f"v1 * 2 = {v1 * 2}")      
    print(f"2 * v1 = {2 * v1}")    
    print(f"v1 * v2 = {v1 * v2}")  

    # 异常测试
    try:
        Vector(-1)
    except ValueError as e:
        print(f"捕获异常: {e}")   

    try:
        v1 < v_zero
    except ValueError as e:
        print(f"捕获异常: {e}")    
```
