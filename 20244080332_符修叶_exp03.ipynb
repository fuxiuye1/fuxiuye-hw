```python
#正则表达式
import re

#1.
s1 = "cat on mat; later: cats and dogs"
res_d = re.findall(r"\d+", s1)
res_s = re.findall(r"\s+", s1)
res_cat_word = re.findall(r"\bcat\b", s1)
print("=====任务1=====")
print(r"\d+ 匹配结果：", res_d)
print(r"\s+ 匹配结果：", res_s)
print(r"\bcat\b 匹配结果：", res_cat_word)

#2.
s2 = "cat dog bird CAT"
res_catdog = re.findall(r"cat|dog", s2)
res1 = re.findall(r"a|aa|aaa", "aaaa")
res2 = re.findall(r"aaa|aa|a", "aaaa")
print("\n=====任务2=====")
print(r"cat|dog 匹配结果：", res_catdog)
print(r"a|aa|aaa 匹配'aaaa'：", res1)
print(r"aaa|aa|a 匹配'aaaa'：", res2)
print("说明：正则分支是【顺序优先】，匹配到前面的就不再往后尝试。")

# 3.
s3 = "ct cat caat caaat"
p1 = r"ca*t"
p2 = r"ca+t"
p3 = r"ca?t"
p4 = r"ca{2}t"
p5 = r"ca{1,2}t"
print("\n=====任务3=====")
print(p1, " 结果：", re.findall(p1, s3), " 含义：a出现0次或多次")
print(p2, " 结果：", re.findall(p2, s3), " 含义：a出现至少1次")
print(p3, " 结果：", re.findall(p3, s3), " 含义：a出现0次或1次")
print(p4, " 结果：", re.findall(p4, s3), " 含义：a严格出现2次")
print(p5, " 结果：", re.findall(p5, s3), " 含义：a出现1~2次（包含边界）")
```

```python
#分组、回溯引用与替换
#1.捕获组提取邮箱用户名、域名
s1 = "联系：alice@example.com ; 备选：bob@test.org"
p1 = r"(\w+)@([\w\.]+)"
res1 = re.findall(p1, s1)
print("=====任务1=====")
print("(用户名, 域名)：", res1)

# 2.后向引用匹配连续重复单词
s2 = "cat cat, dog cat, bird bird"
p2 = r"\b(\w+)\s+\1\b"
res2 = re.findall(p2, s2)
print("\n=====任务2=====")
print("连续重复单词：", res2)

#3.日期格式转换 年-月-日 → 月-日-年
s3 = "2024-05-01\n2024/05/02\n2024.05.03"
p3 = r"(\d{4})[-/.](\d{2})[-/.](\d{2})"
s3_new = re.sub(p3, r"\2-\3-\1", s3)
print("\n=====任务3=====")
print("转换前：")
print(s3)
print("转换后：")
print(s3_new)

#4.电话号码统一格式 (123) 456-7890
s4 = "联系我：123-456-7890 或 987.654.3210"
p4 = r"(\d{3})[-.](\d{3})[-.](\d{4})"
s4_new = re.sub(p4, r"(\1) \2-\3", s4)
print("\n=====任务4=====")
print("转换前：", s4)
print("转换后：", s4_new)
```

```python
#解析HTML
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")

# 1.
print("=====任务1=====")
title_text = soup.title.get_text()
h1_tag = soup.h1
print("页面标题文本：", title_text)
print("h1标签名：", h1_tag.name)
print("h1文本：", h1_tag.get_text())

#2.
print("\n=====任务2=====")
p_list = soup.find_all("p", class_="course")
print("所有class='course'的段落：")
for p in p_list:
    print("  ", p.get_text())

a_list = soup.find_all("a")
print("所有链接：")
for a in a_list:
    link_text = a.get_text()
    link_href = a["href"]
    print(f"  链接文字：{link_text}, href：{link_href}")

#3.
print("\n=====任务3=====")
doc_text = soup.get_text()
single_p_text = p_list[0].get_text()
single_a_text = a_list[0].get_text()
print("整个文档get_text()：")
print(doc_text)
print("第一个段落get_text()：", single_p_text)
print("第一个链接get_text()：", single_a_text)
```
