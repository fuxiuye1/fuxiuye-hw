```python
#欧几里得算法
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

# 测试三组数字
print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
#负数
print(gcd(-12, -8))
print(gcd(-12, 8))
```

```python
#近似欧拉数e
import math

# 1. euler_limit：用 (1+1/n)^n 近似e
def euler_limit(n):
    return (1 + 1 / n) ** n

# 2. euler_infinite_sum：前n项求和，公式 e = sum(1/k!)，k从0开始
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1  # 0! = 1
    for k in range(n):
        total += 1 / fact
        fact *= (k + 1)
    return total

# 3. euler_approx：循环累加，直到误差小于epsilon
def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(true_e - total) > epsilon:
        total += 1 / fact
        k += 1
        fact *= k
    return total

# 4. print_euler_sum_table：打印k从1~n的euler_infinite_sum(k)
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

# 测试第2个函数示例
print(euler_infinite_sum(0))  
print(euler_infinite_sum(1))  
print(euler_infinite_sum(4))   

# 测试euler_limit
print(euler_limit(10000))

# 测试euler_approx，误差小于0.001
print(euler_approx(0.001))

# 打印表格
print_euler_sum_table(5)
```

```python
#测试整数属性
#  1. 判断是否是2的幂
def is_power_of_2(n):
    # 不断除以2，直到不能整除
    while n % 2 == 0:
        n = n // 2
    # 最后剩下等于1，就是2的幂
    return n == 1

# 2. 推广：判断n是不是b的幂
def is_power(b, n):
    # 特殊：任何数的0次方=1，所以n=1一定返回True
    while n % b == 0:
        n = n // b
    return n == 1

# is_power_of_2测试
print(is_power_of_2(1))   
print(is_power_of_2(8))   
print(is_power_of_2(6))   
print(is_power_of_2(16)) 

# is_power(b,n)测试
print(is_power(2, 8))    
print(is_power(3,9))    
print(is_power(5,125))   
print(is_power(2,10))     
print(is_power(7,1))     
```
