```python
# 实验2
# ========== 任务1：分数等级函数 ==========
def score_level(score):
    # 判断分数是否在合法区间
    if not 0 <= score <= 100:
        raise ValueError("分数超出0~100范围")
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"


# ========== 任务2：for、while求1~100和 ==========
# for循环
s_for = 0
for i in range(1, 101):
    s_for += i
print("for循环求和：", s_for)

# while循环
s_while, i = 0, 1
while i <= 100:
    s_while += i
    i += 1
print("while循环求和：", s_while)


# ========== 任务3：迭代版、递归版阶乘 ==========
# 迭代阶乘
def fact_iter(n):
    res = 1
    for i in range(2, n+1):
        res *= i
    return res

# 递归阶乘
def fact_rec(n):
    if n <= 1:  # 终止条件 0! = 1，1! =1
        return 1
    return n * fact_rec(n - 1)

# 测试 n=0,1,5,10
for n in [0, 1, 5, 10]:
    print(f"n={n},迭代:{fact_iter(n)},递归:{fact_rec(n)}")


# ========== 任务4：summary返回最小、最大、平均值 ==========
def summary(values):
    min_v = min(values)
    max_v = max(values)
    avg_v = sum(values) / len(values)
    return min_v, max_v, avg_v

# 元组解包
res = summary([1,2,3,4])
mi, ma, avg = res
print("最小值，最大值，平均值：", mi, ma, avg)


# ========== 任务5：可变参数求和函数 sum_all(*args) ==========
def sum_all(*args):
    # *args接收任意多个位置参数，打包成元组
    return sum(args)

# 测试无参、单个、多个参数
print("无参：", sum_all())
print("单个参数：", sum_all(5))
print("多个参数：", sum_all(1,2,3))
```

```python
#实验4
import os
import pickle

# =========任务1：创建目录=========
print("当前工作目录：", os.getcwd())
data_dir = "exp01_data"
os.makedirs(data_dir, exist_ok=True)
file_path = os.path.join(data_dir, "students.txt")

# =========任务2：写入并读取文件=========
text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
# 写入
with open(file_path, "w", encoding="utf-8") as f:
    f.write(text)

# 读取，跳过表头
with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()[1:]

# =========任务3：解析函数=========
def parse_student(line):
    line = line.strip()
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError("字段数量不对")
    name, score_str = parts
    if not name.strip():
        raise ValueError("姓名不能为空")
    score = int(score_str)
    if not 0 <= score <= 100:
        raise ValueError("分数不在0~100范围")
    return {"name":name, "score":score}

# =========任务4：逐行处理捕获异常=========
valid_list = []
valid_cnt = 0
invalid_cnt = 0
err_log = []
for i, line in enumerate(lines, start=2): #行号从2（跳过表头）
    try:
        s = parse_student(line)
        valid_list.append(s)
        valid_cnt +=1
    except Exception as e:
        invalid_cnt +=1
        err_log.append(f"行{i} 错误:{e}")

# 统计平均分、及格率
scores = [x["score"] for x in valid_list]
avg = sum(scores)/len(scores)
pass_cnt = sum(1 for s in scores if s>=60)
pass_rate = pass_cnt / len(scores)

# =========任务5：写入report.txt，演示w和a=========
report_path = os.path.join(data_dir, "report.txt")
# w模式：覆盖原有内容
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录:{valid_cnt}\n无效记录:{invalid_cnt}\n平均分:{avg:.2f}\n及格率:{pass_rate:.2%}\n")
# a模式：追加内容，不会覆盖
with open(report_path, "a", encoding="utf-8") as f:
    f.write("【追加内容】a模式不会清空旧文本\n")

# =========任务6：pickle保存读取=========
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_list, f)
with open(pkl_path, "rb") as f:
    load_data = pickle.load(f)
print("pickle读取是否相等：", load_data == valid_list)

# =========任务7：把parse_student写入student_utils.py（生成模块文件）=========
mod_text = '''
def parse_student(line):
    line = line.strip()
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError("字段数量不对")
    name, score_str = parts
    if not name.strip():
        raise ValueError("姓名不能为空")
    score = int(score_str)
    if not 0 <= score <=100:
        raise ValueError("分数不在0~100范围")
    return {"name":name, "score":score}
'''
with open("student_utils.py","w",encoding="utf-8") as f:
    f.write(mod_text)

# 导入模块测试
from student_utils import parse_student
# 有效、无效测试
try:
    print("有效测试：", parse_student("Tom,77"))
except Exception as e:
    print(e)
try:
    parse_student("Tom,108")
except Exception as e:
    print("无效测试报错：",e)

# =========任务8：捕获文件不存在异常=========
try:
    with open("not_exist.txt","r") as f:
        f.read()
except FileNotFoundError:
    print("捕获异常：文件不存在！")
```

```python
#实验5
import copy

# 任务1：定义Student类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    def is_passed(self):
        return self.score >= 60
    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"

# 任务2：创建Student对象（沿用实验4有效数据）
stu_data = [
    {"name":"Alice","score":85},
    {"name":"Bob","score":58},
    {"name":"Carol","score":92},
    {"name":"David","score":76},
    {"name":"Eva","score":60},
    {"name":"Frank","score":89}
]
stu_objs = []
for d in stu_data:
    s = Student(d["name"], d["score"])
    stu_objs.append(s)
    print(s, "及格" if s.is_passed() else "不及格")

# 任务3：子类GraduateStudent继承Student
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic
    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, topic={self.research_topic})"
# 测试子类
g = GraduateStudent("Tom",88,"大数据分析")
print(g, g.is_passed())

# 任务4：浅拷贝、深拷贝对比
orig = {"name": "Alice", "scores": [85, 90]}
# 直接赋值
a1 = orig
a1["scores"][0] = 99
print("直接赋值 orig:", orig)
orig = {"name": "Alice", "scores": [85, 90]} #重置
# copy浅拷贝
a2 = copy.copy(orig)
a2["scores"][0] = 99
print("copy.copy  orig:", orig)
orig = {"name": "Alice", "scores": [85, 90]} #重置
# deepcopy深拷贝
a3 = copy.deepcopy(orig)
a3["scores"][0] = 99
print("deepcopy orig:", orig)
```
